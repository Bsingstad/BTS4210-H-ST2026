### Oppgave 2 – Del dataene og lag en baseline

Del i trening (80 %) og test (20 %) med random_state=0. Beregn RMSE, MAE og R2 på testsettet for en baseline-modell (dummy) som alltid gjetter gjennomsnittet av treningsprisene.

```python
y = dia["price"]
X = dia.drop(columns="price")
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2, random_state=0)

def rapport(y_sann, y_pred, navn=""):
    rmse = np.sqrt(mean_squared_error(y_sann, y_pred))
    mae = mean_absolute_error(y_sann, y_pred)
    print(f"{navn:28s} RMSE = {rmse:8.0f}  MAE = {mae:8.0f}  R^2 = {r2_score(y_sann, y_pred):.3f}  MAE/snittpris = {mae / y_sann.mean():.1%}")

base = DummyRegressor().fit(X_train, y_train)
rapport(y_test, base.predict(X_test), "Baseline (gjennomsnitt)")
```

**Løsningsforslag – svar:**


Baseline gir RMSE ca lik 4 000 USD og R2 ca lik 0 (som forventet mtp estimering basert på gjennomsnitt). Dette er tallet alle modeller bør slå.

### Oppgave 3 – Lineær regresjon: original vs. logaritmisk skala
- a) Tren LinearRegression på de numeriske variablene  for å predikere `price`. Rapporter testmetrikker.
- b) Tren samme modell for å log-normlisere skjevfordelte variabler (også target). Regn prediksjonene tilbake med np.exp og rapporter RMSE/MAE i USD.
- c) Hvilken metode er best? 

```python
num = ["carat", "depth", "table", "x", "y", "z"]
lr = LinearRegression().fit(X_train[num], y_train)
rapport(y_test, lr.predict(X_test[num]), "Lineær (6 numeriske)") # bruker funksjonen fra forrige oppgave

lr_log = LinearRegression().fit(np.log(X_train[["carat"]]), np.log(y_train))
pred_log = np.exp(lr_log.predict(np.log(X_test[["carat"]]))) #transformerer tilbake etter estimering
rapport(y_test, pred_log, "log(pris) --> log(carat)")
```

Logaritmisk med en variabel gir litt lavere MAE (bedre på de typiske, billigere diamantene) men høyere RMSE (dårligere på de få dyre, der en prosentfeil gir store utslag i pris). Valg av metrikk påvirker hvilken modell som er best.

### Oppgave 4 – Pipeline, modellvalg og hyperparametere
Implementer en ColumnTransformer som skalerer de numeriske variablene og one-hot-encode de kategoriske. Bruk y = `log(price)` som target. Sammenlign med 5-fold kryssvalidering (RMSE på log-skala) på treningsdata:
Ridge, Lasso(alpha=0.001, max_iter=20000) og RandomForestRegressor(n_estimators=100, random_state=0).
Gjør deretter et grid search (GridSearchCV() for den beste modellen (minst 2 hyperparametere).

```python
num = ["carat", "depth", "table", "x", "y", "z"]; kat = ["cut", "color", "clarity"]
prep = ColumnTransformer([("num", StandardScaler(), num), ("kat", OneHotEncoder(handle_unknown="ignore"), kat)])
ylog_train, ylog_test = np.log(y_train), np.log(y_test)
kandidater = {"Ridge": Ridge(1.0), "Lasso": Lasso(alpha=0.001, max_iter=20000), "Random forest": RandomForestRegressor(100, random_state=0)}
cv = KFold(5, shuffle=True, random_state=0)
for navn, m in kandidater.items():
    s = -cross_val_score(Pipeline([("prep", prep), ("m", m)]), X_train, ylog_train, cv=cv, scoring="neg_root_mean_squared_error")
    print(f"{navn:14s} CV-RMSE (log) = {s.mean():.4f} ± {s.std():.4f}")

pipe = Pipeline([("prep", prep), ("m", RandomForestRegressor(random_state=0))])
gs = GridSearchCV(pipe, {"m__n_estimators": [100, 200], "m__max_depth": [10, None], "m__min_samples_leaf": [1, 3]},
                  cv=cv, scoring="neg_root_mean_squared_error").fit(X_train, ylog_train)
print("Beste:", gs.best_params_, " CV-RMSE:", round(-gs.best_score_, 4))
```

Random forest ser tilsyneslatende ut til å gi best og mest stabil ytelse fordi med lavere RMSE og lavere SD. Optimalisering med grid search gir bare marginale forbedringer (standardparametere er ofte gode for random forest).

### Oppgave 5 – Endelig test og feilanalyse
- a) Evaluer den beste modellen på testsettet. Rapporter RMSE, MAE og R2 i USD (regn tilbake med exp) og MAE i forhold til gjennomsnittlig pris.
- b) Plott predikert vs. faktisk pris, og residualer vs. predikert pris.
- c) Del testsettet i tre karat-grupper (< 0,5 / 0,5–1,0 / > 1,0). Beregn RMSE, MAE og R2 per gruppe. Hvor er modellen dårligst?

```python
beste = gs.best_estimator_
pred = np.exp(beste.predict(X_test))
rapport(y_test, pred, "Random forest (hyperparametertunet), test")

fig, ax = plt.subplots(1, 2, figsize=(12, 4.5))
ax[0].scatter(y_test, pred, s=8, alpha=0.5); lim = [0, y_test.max()]
ax[0].plot(lim, lim, "k--")
ax[0].set_xlabel("Faktisk pris")
ax[0].set_ylabel("Predikert pris")

ax[1].scatter(pred, y_test - pred, s=8, alpha=0.5)
ax[1].axhline(0, color="red")
ax[1].set_xlabel("Predikert pris")
ax[1].set_ylabel("Residual (USD)")

plt.tight_layout()
plt.show()

gr = pd.cut(X_test["carat"], [0, 0.5, 1.0, 10], labels=["< 0.5", "0.5–1.0", "> 1.0"])
rel = (np.abs(y_test - pred) / y_test).groupby(gr, observed=True).median()
print("Median relativ feil per karatgruppe:\n", rel.round(3))
```

Ytelse langt bedre enn de lineære modellene i oppgave 3. I absolutte tall er feilene størst for dyre, store diamanter, men relativt sett er median prosentfeil høyest for de minste/billigste (ca 8 % for < 0,5 karat mot ca 6–7 % for de andre). Hvilken feilmetrikk som er "riktig» avhenger av bruken: en forhandler bryr seg kanskje om prosentfeil, en forsikringsselskap om absolutt kostnadsfeil.
